# SVM (Support Vector Machine)

## 1.Importing libraries and classes and reading the file

```python
from sklearn.pipeline import Pipeline
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
import pandas as pd
df=pd.read_csv("/kaggle/input/breast-cancer-wisconsin-data/data.csv")
```

```python
#drop useless columns
df=df.drop("id" , axis=1)
df=df.drop("Unnamed: 32" , axis=1)
```

```python
#checking the size of the dataset
df.shape
```

```python
#check for missing values
df.isnull().sum()
```

```python
#check data types
df.info()
```

## 2.Define X as features and Y as a target

```python
X=df.drop("diagnosis" , axis=1)
Y=df["diagnosis"]
```

## 3.split X and Y into train and test(80% of train and 20% of test for each)


```python
X_Train, X_Test, Y_Train, Y_Test = train_test_split(
    X, Y, test_size=0.2, random_state=42, shuffle=True
)
```

## 4.Setting a pipeline in which we scale the features then training the model

```python

model = Pipeline([     ("scaler", StandardScaler()),     ("svm", SVC()) ])
model.fit(X_Train, Y_Train)
```

## 5.Predicting the Y_test

```python
prediction=model.predict(X_Test)
```

## 6.printing the report of the model

```python
print(classification_report(Y_Test,prediction))
```

# KNN(K-Nearest Neighbors)

```python
knn=Pipeline([  ("scaler2",StandardScaler()) , ("knn" ,KNeighborsClassifier(n_neighbors=8) )])
knn.fit(X_Train , Y_Train)
```

```python
knn_pred=knn.predict(X_Test)
```

```python
print(classification_report(Y_Test , knn_pred))
```

# DecisionTree

```python
DT=Pipeline([ ("DT" ,DecisionTreeClassifier(random_state=350) )])
DT.fit(X_Train , Y_Train)
```

```python
DT_pred=DT.predict(X_Test)
```

```python
print(classification_report(Y_Test , DT_pred) )
```

# RandomForest

```python
RF=Pipeline([ ("RF" ,RandomForestClassifier(random_state=425) )])
RF.fit(X_Train , Y_Train)
```

```python
RF_pred=RF.predict(X_Test)
```

```python
print(classification_report(Y_Test , RF_pred) )
```
